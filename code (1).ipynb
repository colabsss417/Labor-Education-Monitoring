```python
# ============================================================
# STEP 11: AJA-EnDNN PROPOSED MODEL
# Adaptive Jaya Optimization + EnDNN
# ============================================================

import numpy as np
import pandas as pd
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    matthews_corrcoef
)

# ============================================================
# 1. DATA
# ============================================================

X = np.asarray(X_train_final, dtype=np.float32)
y = np.asarray(y_train_final, dtype=np.int32)

X_test = np.asarray(X_test_final, dtype=np.float32)
y_test = np.asarray(y_test_final, dtype=np.int32)

INPUT_DIM = X.shape[1]
NUM_CLASSES = num_classes


# ============================================================
# 2. METRIC FUNCTION
# ============================================================

def calculate_metrics(y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Multiclass sensitivity and specificity
    # --------------------------------------------------------
    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=np.arange(NUM_CLASSES)
    )

    sensitivities = []
    specificities = []

    for i in range(NUM_CLASSES):

        TP = cm[i, i]

        FN = np.sum(cm[i, :]) - TP

        FP = np.sum(cm[:, i]) - TP

        TN = np.sum(cm) - (
            TP + FN + FP
        )

        sensitivity = (
            TP / (TP + FN)
            if (TP + FN) > 0
            else 0
        )

        specificity = (
            TN / (TN + FP)
            if (TN + FP) > 0
            else 0
        )

        sensitivities.append(sensitivity)
        specificities.append(specificity)

    sensitivity = np.mean(sensitivities)
    specificity = np.mean(specificities)

    # --------------------------------------------------------
    # MCC
    # --------------------------------------------------------
    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )

    return {
        "Accuracy (%)": accuracy * 100,
        "Precision (%)": precision * 100,
        "Recall (%)": recall * 100,
        "F1-Score (%)": f1 * 100,
        "Sensitivity (%)": sensitivity * 100,
        "Specificity (%)": specificity * 100,
        "MCC": mcc
    }


# ============================================================
# 3. ENDDNN MODEL BUILDER
# ============================================================

def build_endnn(
    input_dim,
    num_classes,
    learning_rate=0.001,
    dropout1=0.30,
    dropout2=0.25,
    dropout3=0.20,
    units1=128,
    units2=64,
    units3=32
):

    model = Sequential([

        Input(
            shape=(input_dim,),
            name="PCA_Input"
        ),

        Dense(
            units1,
            activation="relu",
            name="Dense_128"
        ),

        BatchNormalization(
            name="BatchNorm_1"
        ),

        Dropout(
            dropout1,
            name="Dropout_1"
        ),

        Dense(
            units2,
            activation="relu",
            name="Dense_64"
        ),

        BatchNormalization(
            name="BatchNorm_2"
        ),

        Dropout(
            dropout2,
            name="Dropout_2"
        ),

        Dense(
            units3,
            activation="relu",
            name="Dense_32"
        ),

        Dropout(
            dropout3,
            name="Dropout_3"
        ),

        Dense(
            num_classes,
            activation="softmax",
            name="Output"
        )
    ])

    model.compile(
        optimizer=Adam(
            learning_rate=learning_rate
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 4. ADAPTIVE JAYA OPTIMIZATION
# ============================================================

# ------------------------------------------------------------
# Hyperparameter search space
# ------------------------------------------------------------

SEARCH_SPACE = {

    "learning_rate": (
        0.0001,
        0.001
    ),

    "dropout1": (
        0.20,
        0.40
    ),

    "dropout2": (
        0.15,
        0.35
    ),

    "dropout3": (
        0.10,
        0.30
    )
}


# ------------------------------------------------------------
# Jaya population
# ------------------------------------------------------------

POPULATION_SIZE = 6
JAYA_ITERATIONS = 5

rng = np.random.default_rng(42)


def create_random_solution():

    return np.array([

        rng.uniform(
            SEARCH_SPACE["learning_rate"][0],
            SEARCH_SPACE["learning_rate"][1]
        ),

        rng.uniform(
            SEARCH_SPACE["dropout1"][0],
            SEARCH_SPACE["dropout1"][1]
        ),

        rng.uniform(
            SEARCH_SPACE["dropout2"][0],
            SEARCH_SPACE["dropout2"][1]
        ),

        rng.uniform(
            SEARCH_SPACE["dropout3"][0],
            SEARCH_SPACE["dropout3"][1]
        )

    ])


population = np.array([
    create_random_solution()
    for _ in range(POPULATION_SIZE)
])


# ------------------------------------------------------------
# Validation split for optimization
# ------------------------------------------------------------

skf_opt = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

train_idx_opt, val_idx_opt = next(
    skf_opt.split(X, y)
)

X_opt_train = X[train_idx_opt]
X_opt_val = X[val_idx_opt]

y_opt_train = y[train_idx_opt]
y_opt_val = y[val_idx_opt]


# ------------------------------------------------------------
# Fitness function
# ------------------------------------------------------------

def evaluate_solution(solution):

    learning_rate = float(solution[0])
    dropout1 = float(solution[1])
    dropout2 = float(solution[2])
    dropout3 = float(solution[3])

    tf.keras.backend.clear_session()

    model = build_endnn(
        INPUT_DIM,
        NUM_CLASSES,
        learning_rate=learning_rate,
        dropout1=dropout1,
        dropout2=dropout2,
        dropout3=dropout3
    )

    model.fit(
        X_opt_train,
        y_opt_train,
        validation_data=(
            X_opt_val,
            y_opt_val
        ),
        epochs=15,
        batch_size=32,
        verbose=0,
        callbacks=[
            EarlyStopping(
                monitor="val_loss",
                patience=4,
                restore_best_weights=True
            )
        ]
    )

    probabilities = model.predict(
        X_opt_val,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    metrics = calculate_metrics(
        y_opt_val,
        predictions
    )

    # --------------------------------------------------------
    # Fitness = validation macro F1
    # --------------------------------------------------------
    fitness = metrics["F1-Score (%)"]

    return fitness


# ============================================================
# 5. ADAPTIVE JAYA ITERATIONS
# ============================================================

best_solution = None
best_fitness = -np.inf

optimization_history = []


for iteration in range(JAYA_ITERATIONS):

    fitness_values = np.array([

        evaluate_solution(solution)

        for solution in population

    ])

    best_index = np.argmax(
        fitness_values
    )

    worst_index = np.argmin(
        fitness_values
    )

    iteration_best = population[
        best_index
    ].copy()

    iteration_worst = population[
        worst_index
    ].copy()

    iteration_fitness = fitness_values[
        best_index
    ]

    if iteration_fitness > best_fitness:

        best_fitness = iteration_fitness

        best_solution = iteration_best.copy()


    # --------------------------------------------------------
    # Adaptive Jaya coefficient
    # --------------------------------------------------------

    progress = (
        iteration /
        max(JAYA_ITERATIONS - 1, 1)
    )

    adaptive_factor = (
        1.0 - 0.5 * progress
    )


    # --------------------------------------------------------
    # Jaya population update
    # --------------------------------------------------------

    new_population = []

    for i in range(POPULATION_SIZE):

        current = population[i]

        r1 = rng.random(
            len(current)
        )

        r2 = rng.random(
            len(current)
        )

        candidate = (

            current

            + adaptive_factor
            * r1
            * (
                iteration_best
                - np.abs(current)
            )

            - adaptive_factor
            * r2
            * (
                iteration_worst
                - np.abs(current)
            )
        )


        # ----------------------------------------------------
        # Boundary control
        # ----------------------------------------------------

        candidate[0] = np.clip(
            candidate[0],
            SEARCH_SPACE["learning_rate"][0],
            SEARCH_SPACE["learning_rate"][1]
        )

        candidate[1] = np.clip(
            candidate[1],
            SEARCH_SPACE["dropout1"][0],
            SEARCH_SPACE["dropout1"][1]
        )

        candidate[2] = np.clip(
            candidate[2],
            SEARCH_SPACE["dropout2"][0],
            SEARCH_SPACE["dropout2"][1]
        )

        candidate[3] = np.clip(
            candidate[3],
            SEARCH_SPACE["dropout3"][0],
            SEARCH_SPACE["dropout3"][1]
        )

        new_population.append(
            candidate
        )


    population = np.array(
        new_population
    )

    optimization_history.append({
        "Iteration": iteration + 1,
        "Best_F1": iteration_fitness
    })


# ============================================================
# 6. BEST AJA PARAMETERS
# ============================================================

best_learning_rate = float(
    best_solution[0]
)

best_dropout1 = float(
    best_solution[1]
)

best_dropout2 = float(
    best_solution[2]
)

best_dropout3 = float(
    best_solution[3]
)


# ============================================================
# 7. BUILD FINAL AJA-EnDNN
# ============================================================

tf.keras.backend.clear_session()

aja_endnn_model = build_endnn(
    INPUT_DIM,
    NUM_CLASSES,

    learning_rate=best_learning_rate,

    dropout1=best_dropout1,
    dropout2=best_dropout2,
    dropout3=best_dropout3
)


# ============================================================
# 8. FINAL TRAINING
# ============================================================

early_stopping_final = EarlyStopping(
    monitor="val_loss",
    patience=10,
    mode="min",
    restore_best_weights=True,
    verbose=0
)

reduce_lr_final = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    mode="min",
    verbose=0
)


aja_history = aja_endnn_model.fit(

    X,
    y,

    validation_split=0.20,

    epochs=50,

    batch_size=32,

    shuffle=True,

    callbacks=[
        early_stopping_final,
        reduce_lr_final
    ],

    verbose=0
)


print("\n" + "=" * 90)
print("AJA-EnDNN TRAINING COMPLETED")
print("=" * 90)
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    BatchNormalization,
    Dropout
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)


print("=" * 80)
print("STEP 11: 5-FOLD CROSS-VALIDATION - SAME EnDNN MODEL")
print("=" * 80)


# ============================================================
# 1. USE EXISTING PCA DATA
# ============================================================

X_cv = np.asarray(
    X_train_final,
    dtype=np.float32
)

# Convert one-hot labels to integer labels
if y_train_onehot.ndim == 2:
    y_cv = np.argmax(
        y_train_onehot,
        axis=1
    )
else:
    y_cv = np.asarray(
        y_train_onehot
    )

num_classes = 4


print("\nDataset Shape:")
print(X_cv.shape)

print("\nTarget Shape:")
print(y_cv.shape)

print("\nNumber of Classes:")
print(num_classes)


# ============================================================
# 2. SAME EnDNN MODEL AS STEP 10
# ============================================================

def build_same_endnn():

    model = Sequential([

        Input(
            shape=(X_train_final.shape[1],),
            name="PCA_Input"
        ),

        Dense(
            128,
            activation="relu",
            name="Dense_128"
        ),

        BatchNormalization(
            name="BatchNorm_1"
        ),

        Dropout(
            0.30,
            name="Dropout_1"
        ),

        Dense(
            64,
            activation="relu",
            name="Dense_64"
        ),

        BatchNormalization(
            name="BatchNorm_2"
        ),

        Dropout(
            0.25,
            name="Dropout_2"
        ),

        Dense(
            32,
            activation="relu",
            name="Dense_32"
        ),

        Dropout(
            0.20,
            name="Dropout_3"
        ),

        Dense(
            num_classes,
            activation="softmax",
            name="Output"
        )
    ])

    # SAME optimizer and loss as original model
    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# ============================================================
# 3. SAME TRAINING PARAMETERS
# ============================================================

EPOCHS = 50
BATCH_SIZE = 32

# SAME EarlyStopping
EARLY_STOPPING_PATIENCE = 10

# SAME ReduceLROnPlateau
LR_FACTOR = 0.5
LR_PATIENCE = 5
MIN_LR = 1e-6


# ============================================================
# 4. 5-FOLD STRATIFIED SPLIT
# ============================================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ============================================================
# 5. RESULT STORAGE
# ============================================================

fold_results = []


# ============================================================
# 6. RUN 5 FOLDS
# ============================================================

for fold, (train_idx, val_idx) in enumerate(
    skf.split(X_cv, y_cv),
    start=1
):

    print("\n")
    print("=" * 80)
    print(f"FOLD {fold}/5")
    print("=" * 80)

    # --------------------------------------------------------
    # Fold data
    # --------------------------------------------------------

    X_fold_train = X_cv[train_idx]
    X_fold_val = X_cv[val_idx]

    y_fold_train = y_cv[train_idx]
    y_fold_val = y_cv[val_idx]

    # --------------------------------------------------------
    # Convert integer labels to one-hot
    # SAME categorical_crossentropy setup as original model
    # --------------------------------------------------------

    y_fold_train_onehot = tf.keras.utils.to_categorical(
        y_fold_train,
        num_classes=num_classes
    )

    y_fold_val_onehot = tf.keras.utils.to_categorical(
        y_fold_val,
        num_classes=num_classes
    )

    print("\nTraining Shape:")
    print(X_fold_train.shape)

    print("\nValidation Shape:")
    print(X_fold_val.shape)

    # --------------------------------------------------------
    # Reset random seeds
    # --------------------------------------------------------

    tf.keras.backend.clear_session()

    tf.random.set_seed(42 + fold)
    np.random.seed(42 + fold)

    # --------------------------------------------------------
    # BUILD EXACT SAME EnDNN
    # --------------------------------------------------------

    fold_model = build_same_endnn()

    # --------------------------------------------------------
    # SAME CALLBACKS AS ORIGINAL TRAINING
    # --------------------------------------------------------

    early_stopping = EarlyStopping(
        monitor="val_loss",
        patience=EARLY_STOPPING_PATIENCE,
        mode="min",
        restore_best_weights=True,
        verbose=0
    )

    reduce_lr = ReduceLROnPlateau(
        monitor="val_loss",
        factor=LR_FACTOR,
        patience=LR_PATIENCE,
        min_lr=MIN_LR,
        mode="min",
        verbose=0
    )

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    fold_history = fold_model.fit(

        X_fold_train,
        y_fold_train_onehot,

        validation_data=(
            X_fold_val,
            y_fold_val_onehot
        ),

        epochs=EPOCHS,

        batch_size=BATCH_SIZE,

        shuffle=True,

        callbacks=[
            early_stopping,
            reduce_lr
        ],

        verbose=0
    )

    # --------------------------------------------------------
    # PREDICTION
    # --------------------------------------------------------

    y_fold_prob = fold_model.predict(
        X_fold_val,
        batch_size=32,
        verbose=0
    )

    y_fold_pred = np.argmax(
        y_fold_prob,
        axis=1
    )

    # ========================================================
    # 7. ACCURACY
    # ========================================================

    accuracy = accuracy_score(
        y_fold_val,
        y_fold_pred
    )


    # ========================================================
    # 8. PRECISION
    # ========================================================

    precision = precision_score(
        y_fold_val,
        y_fold_pred,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # 9. RECALL
    # ========================================================

    recall = recall_score(
        y_fold_val,
        y_fold_pred,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # 10. F1 SCORE
    # ========================================================

    f1 = f1_score(
        y_fold_val,
        y_fold_pred,
        average="macro",
        zero_division=0
    )


    # ========================================================
    # 11. CONFUSION MATRIX
    # ========================================================

    cm = confusion_matrix(
        y_fold_val,
        y_fold_pred,
        labels=np.arange(num_classes)
    )


    # ========================================================
    # 12. SENSITIVITY + SPECIFICITY
    # ========================================================

    sensitivity_values = []
    specificity_values = []

    for class_id in range(num_classes):

        TP = cm[class_id, class_id]

        FN = (
            np.sum(cm[class_id, :])
            - TP
        )

        FP = (
            np.sum(cm[:, class_id])
            - TP
        )

        TN = (
            np.sum(cm)
            - TP
            - FN
            - FP
        )

        # Sensitivity
        if (TP + FN) > 0:
            sensitivity_class = (
                TP / (TP + FN)
            )
        else:
            sensitivity_class = 0.0

        # Specificity
        if (TN + FP) > 0:
            specificity_class = (
                TN / (TN + FP)
            )
        else:
            specificity_class = 0.0

        sensitivity_values.append(
            sensitivity_class
        )

        specificity_values.append(
            specificity_class
        )


    # Macro sensitivity
    sensitivity = np.mean(
        sensitivity_values
    )

    # Macro specificity
    specificity = np.mean(
        specificity_values
    )


    # ========================================================
    # 13. MCC
    # ========================================================

    mcc = matthews_corrcoef(
        y_fold_val,
        y_fold_pred
    )


    # ========================================================
    # 14. BEST VALIDATION RESULTS
    # ========================================================

    best_epoch = (
        np.argmin(
            fold_history.history["val_loss"]
        ) + 1
    )

    best_val_loss = np.min(
        fold_history.history["val_loss"]
    )

    best_val_accuracy = np.max(
        fold_history.history["val_accuracy"]
    )


    # ========================================================
    # 15. STORE FOLD RESULTS
    # ========================================================

    fold_results.append({

        "Fold": fold,

        "Accuracy": accuracy,

        "Precision": precision,

        "Recall": recall,

        "F1 Score": f1,

        "Sensitivity": sensitivity,

        "Specificity": specificity,

        "MCC": mcc,

        "Best Epoch": best_epoch,

        "Best Val Loss": best_val_loss,

        "Best Val Accuracy": best_val_accuracy

    })


    # ========================================================
    # 16. PRINT FOLD RESULT
    # ========================================================

    print("\nFold Results:")

    print(
        f"Accuracy      : {accuracy:.6f}"
    )

    print(
        f"Precision     : {precision:.6f}"
    )

    print(
        f"Recall        : {recall:.6f}"
    )

    print(
        f"F1 Score      : {f1:.6f}"
    )

    print(
        f"Sensitivity   : {sensitivity:.6f}"
    )

    print(
        f"Specificity   : {specificity:.6f}"
    )

    print(
        f"MCC           : {mcc:.6f}"
    )

    print(
        f"Best Epoch    : {best_epoch}"
    )

    print(
        f"Best Val Loss : {best_val_loss:.6f}"
    )

    print(
        f"Best Val Acc. : {best_val_accuracy:.6f}"
    )


# ============================================================
# 17. CREATE RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(
    fold_results
)


# ============================================================
# 18. CALCULATE OVERALL MEAN
# ============================================================

metric_columns = [

    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "Sensitivity",
    "Specificity",
    "MCC"

]

mean_values = results_df[
    metric_columns
].mean()


# ============================================================
# 19. ADD OVERALL MEAN ROW
# ============================================================

mean_row = {

    "Fold": "Overall Mean",

    "Accuracy":
        mean_values["Accuracy"],

    "Precision":
        mean_values["Precision"],

    "Recall":
        mean_values["Recall"],

    "F1 Score":
        mean_values["F1 Score"],

    "Sensitivity":
        mean_values["Sensitivity"],

    "Specificity":
        mean_values["Specificity"],

    "MCC":
        mean_values["MCC"],

    "Best Epoch":
        results_df["Best Epoch"].mean(),

    "Best Val Loss":
        results_df["Best Val Loss"].mean(),

    "Best Val Accuracy":
        results_df["Best Val Accuracy"].mean()
}


results_final = pd.concat(
    [
        results_df,
        pd.DataFrame([mean_row])
    ],
    ignore_index=True
)

# ============================================================
# 20. OVERALL MEAN ONLY
# ============================================================

print("\n")
print("=" * 80)
print("OVERALL MEAN")
print("=" * 80)

print(
    f"Accuracy      : "
    f"{mean_values['Accuracy']:.6f}"
)

print(
    f"Precision     : "
    f"{mean_values['Precision']:.6f}"
)

print(
    f"Recall        : "
    f"{mean_values['Recall']:.6f}"
)

print(
    f"F1 Score      : "
    f"{mean_values['F1 Score']:.6f}"
)

print(
    f"Sensitivity   : "
    f"{mean_values['Sensitivity']:.6f}"
)

print(
    f"Specificity   : "
    f"{mean_values['Specificity']:.6f}"
)

print(
    f"MCC           : "
    f"{mean_values['MCC']:.6f}"
)

print("=" * 80)
```

```python
# ============================================================
# FULL ABLATION FLOW
# Baseline DNN
# + Data Preprocessing
# + PCA Feature Extraction
# + EnDNN
# + AJA Optimization
# + Edge Computing Integration
# AJA-EnDNN [Proposed]
# ============================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout,
    BatchNormalization
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau
)

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

# ============================================================
# 2. DATASET PATH
# ============================================================

DATA_PATH = r"C:\Users\SIRPI\Downloads\Edge_Labor_Education_Dataset.csv"

# ------------------------------------------------------------
# IMPORTANT:
# Change DATA_PATH above if your original CSV is stored
# somewhere else.
# ------------------------------------------------------------

print("=" * 100)
print("STEP 1: LOAD DATASET")
print("=" * 100)

df = pd.read_csv(DATA_PATH)

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

# ============================================================
# 3. TARGET CONFIGURATION
# ============================================================

# Based on your previous output:
#
# 0 -> High
# 1 -> Low
# 2 -> Medium
# 3 -> Very Low
#
# Therefore Overall_Performance_Category is used as target.

TARGET_COLUMN = "Overall_Performance_Category"

if TARGET_COLUMN not in df.columns:
    raise ValueError(
        f"Target column '{TARGET_COLUMN}' was not found in the dataset."
    )

print("\nTarget Column:")
print(TARGET_COLUMN)

print("\nOriginal Target Distribution:")
print(df[TARGET_COLUMN].value_counts())

# ============================================================
# 4. SEPARATE FEATURES AND TARGET
# ============================================================

print("\n" + "=" * 100)
print("STEP 2: SEPARATE FEATURES AND TARGET")
print("=" * 100)

y_raw = df[TARGET_COLUMN].copy()

X_raw = df.drop(
    columns=[TARGET_COLUMN]
).copy()

# ------------------------------------------------------------
# Remove non-predictive ID/date/object columns automatically
# ------------------------------------------------------------

drop_columns = []

for col in X_raw.columns:

    if (
        X_raw[col].dtype == "object"
        or
        "date" in col.lower()
        or
        "id" in col.lower()
    ):
        drop_columns.append(col)

if len(drop_columns) > 0:

    print("\nRemoved non-numeric/non-predictive columns:")
    print(drop_columns)

    X_raw = X_raw.drop(
        columns=drop_columns
    )

# ------------------------------------------------------------
# Convert all remaining features to numeric
# ------------------------------------------------------------

X_raw = X_raw.apply(
    pd.to_numeric,
    errors="coerce"
)

# Fill missing values using training-independent median
# only for initial dataset preparation.
X_raw = X_raw.fillna(
    X_raw.median(numeric_only=True)
)

print("\nFeature Shape:")
print(X_raw.shape)

# ============================================================
# 5. ENCODE TARGET
# ============================================================

print("\n" + "=" * 100)
print("STEP 3: TARGET ENCODING")
print("=" * 100)

label_encoder = LabelEncoder()

y = label_encoder.fit_transform(
    y_raw
)

class_names = label_encoder.classes_

num_classes = len(
    class_names
)

print("\nNumber of Classes:")
print(num_classes)

print("\nClass Encoding:")

for i, cls in enumerate(class_names):
    print(
        f"{i} -> {cls}"
    )

# ============================================================
# 6. STRATIFIED TRAIN / TEST SPLIT
# ============================================================

print("\n" + "=" * 100)
print("STEP 4: STRATIFIED TRAIN / TEST SPLIT")
print("=" * 100)

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw,
    y,
    test_size=0.20,
    stratify=y,
    random_state=SEED
)

print("\nTraining Shape:")
print(X_train_raw.shape)

print("\nTesting Shape:")
print(X_test_raw.shape)

print("\nTraining Target Distribution:")
print(
    pd.Series(y_train).value_counts().sort_index()
)

print("\nTesting Target Distribution:")
print(
    pd.Series(y_test).value_counts().sort_index()
)

# ============================================================
# 7. COMMON EVALUATION FUNCTION
# ============================================================

def calculate_metrics(
    y_true,
    y_pred
):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    # --------------------------------------------------------
    # Multiclass sensitivity
    # --------------------------------------------------------

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=np.arange(num_classes)
    )

    sensitivities = []

    specificities = []

    for i in range(num_classes):

        TP = cm[i, i]

        FN = np.sum(cm[i, :]) - TP

        FP = np.sum(cm[:, i]) - TP

        TN = np.sum(cm) - (
            TP + FN + FP
        )

        sensitivity_i = (
            TP / (TP + FN)
            if (TP + FN) > 0
            else 0
        )

        specificity_i = (
            TN / (TN + FP)
            if (TN + FP) > 0
            else 0
        )

        sensitivities.append(
            sensitivity_i
        )

        specificities.append(
            specificity_i
        )

    sensitivity = np.mean(
        sensitivities
    )

    specificity = np.mean(
        specificities
    )

    mcc = matthews_corrcoef(
        y_true,
        y_pred
    )

    return {

        "Accuracy (%)":
            accuracy * 100,

        "Precision (%)":
            precision * 100,

        "Recall (%)":
            recall * 100,

        "F1-Score (%)":
            f1 * 100,

        "Sensitivity (%)":
            sensitivity * 100,

        "Specificity (%)":
            specificity * 100,

        "MCC":
            mcc
    }

# ============================================================
# 8. COMMON DNN PREDICTION FUNCTION
# ============================================================

def train_standard_dnn(
    X_train,
    X_test,
    y_train,
    y_test,
    epochs=30,
    batch_size=32
):

    model = Sequential([

        Input(
            shape=(X_train.shape[1],)
        ),

        Dense(
            128,
            activation="relu"
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            num_classes,
            activation="softmax"
        )
    ])

    model.compile(
        optimizer=Adam(
            learning_rate=0.001
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=7,
        restore_best_weights=True,
        verbose=0
    )

    model.fit(
        X_train,
        y_train,
        validation_split=0.20,
        epochs=epochs,
        batch_size=batch_size,
        shuffle=True,
        callbacks=[early_stop],
        verbose=0
    )

    probabilities = model.predict(
        X_test,
        batch_size=32,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    metrics = calculate_metrics(
        y_test,
        predictions
    )

    return model, metrics

# ============================================================
# 9. BASELINE DNN
# ============================================================

print("\n" + "=" * 100)
print("STEP 5: BASELINE DNN")
print("=" * 100)

baseline_model, baseline_metrics = train_standard_dnn(
    X_train_raw.values.astype("float32"),
    X_test_raw.values.astype("float32"),
    y_train,
    y_test
)

print("\nBaseline DNN Results:")

for key, value in baseline_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 10. DATA PREPROCESSING
# ============================================================

print("\n" + "=" * 100)
print("STEP 6: DATA PREPROCESSING")
print("=" * 100)

scaler = MinMaxScaler(
    feature_range=(0, 1)
)

X_train_scaled = scaler.fit_transform(
    X_train_raw
)

X_test_scaled = scaler.transform(
    X_test_raw
)

print("\nTraining Shape:")
print(X_train_scaled.shape)

print("\nTesting Shape:")
print(X_test_scaled.shape)

print("\nMinimum:")
print(
    X_train_scaled.min()
)

print("\nMaximum:")
print(
    X_train_scaled.max()
)

# ============================================================
# 11. DNN + DATA PREPROCESSING
# ============================================================

print("\n" + "=" * 100)
print("STEP 7: DNN + DATA PREPROCESSING")
print("=" * 100)

preprocessed_model, preprocessing_metrics = train_standard_dnn(
    X_train_scaled.astype("float32"),
    X_test_scaled.astype("float32"),
    y_train,
    y_test
)

print("\nData Preprocessing Results:")

for key, value in preprocessing_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 12. PCA FEATURE EXTRACTION
# ============================================================

print("\n" + "=" * 100)
print("STEP 8: PCA FEATURE EXTRACTION")
print("=" * 100)

# ------------------------------------------------------------
# Retain 95% variance
# ------------------------------------------------------------

pca = PCA(
    n_components=0.95,
    svd_solver="full"
)

X_train_pca = pca.fit_transform(
    X_train_scaled
)

X_test_pca = pca.transform(
    X_test_scaled
)

print("\nOriginal Features:")
print(
    X_train_scaled.shape[1]
)

print("\nPCA Features:")
print(
    X_train_pca.shape[1]
)

print("\nRetained Variance:")
print(
    f"{np.sum(pca.explained_variance_ratio_) * 100:.4f}%"
)

# ============================================================
# 13. PCA + DNN
# ============================================================

print("\n" + "=" * 100)
print("STEP 9: PCA FEATURE EXTRACTION + DNN")
print("=" * 100)

pca_model, pca_metrics = train_standard_dnn(
    X_train_pca.astype("float32"),
    X_test_pca.astype("float32"),
    y_train,
    y_test
)

print("\nPCA-DNN Results:")

for key, value in pca_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 14. EnDNN
# ============================================================

print("\n" + "=" * 100)
print("STEP 10: EnDNN")
print("=" * 100)

# ------------------------------------------------------------
# EXACT EnDNN ARCHITECTURE FROM YOUR PREVIOUS CODE
# ------------------------------------------------------------

tf.random.set_seed(SEED)
np.random.seed(SEED)

endnn_model = Sequential([

    Input(
        shape=(X_train_pca.shape[1],),
        name="PCA_Input"
    ),

    Dense(
        128,
        activation="relu",
        name="Dense_128"
    ),

    BatchNormalization(
        name="BatchNorm_1"
    ),

    Dropout(
        0.30,
        name="Dropout_1"
    ),

    Dense(
        64,
        activation="relu",
        name="Dense_64"
    ),

    BatchNormalization(
        name="BatchNorm_2"
    ),

    Dropout(
        0.25,
        name="Dropout_2"
    ),

    Dense(
        32,
        activation="relu",
        name="Dense_32"
    ),

    Dropout(
        0.20,
        name="Dropout_3"
    ),

    Dense(
        num_classes,
        activation="softmax",
        name="Output"
    )
])

endnn_model.compile(
    optimizer=Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=10,
    mode="min",
    restore_best_weights=True,
    verbose=0
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_lr=1e-6,
    mode="min",
    verbose=0
)

endnn_history = endnn_model.fit(
    X_train_pca,
    y_train,
    validation_split=0.20,
    epochs=50,
    batch_size=32,
    shuffle=True,
    callbacks=[
        early_stopping,
        reduce_lr
    ],
    verbose=0
)

endnn_prob = endnn_model.predict(
    X_test_pca,
    batch_size=32,
    verbose=0
)

endnn_pred = np.argmax(
    endnn_prob,
    axis=1
)

endnn_metrics = calculate_metrics(
    y_test,
    endnn_pred
)

print("\nEnDNN Results:")

for key, value in endnn_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 15. AJA OPTIMIZATION
# ============================================================

print("\n" + "=" * 100)
print("STEP 11: AJA HYPERPARAMETER OPTIMIZATION")
print("=" * 100)

# ------------------------------------------------------------
# Adaptive Jaya Optimization
#
# Parameters optimized:
#   1. Learning rate
#   2. Dropout 1
#   3. Dropout 2
#   4. Dropout 3
#
# The best configuration is selected using validation
# accuracy.
# ------------------------------------------------------------

X_aja_train, X_aja_val, y_aja_train, y_aja_val = train_test_split(
    X_train_pca,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=SEED
)

def build_aja_endnn(
    learning_rate,
    dropout1,
    dropout2,
    dropout3
):

    model = Sequential([

        Input(
            shape=(X_train_pca.shape[1],)
        ),

        Dense(
            128,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            dropout1
        ),

        Dense(
            64,
            activation="relu"
        ),

        BatchNormalization(),

        Dropout(
            dropout2
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dropout(
            dropout3
        ),

        Dense(
            num_classes,
            activation="softmax"
        )
    ])

    model.compile(
        optimizer=Adam(
            learning_rate=learning_rate
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


def aja_objective(position):

    learning_rate = position[0]

    dropout1 = position[1]

    dropout2 = position[2]

    dropout3 = position[3]

    # Ensure valid dropout relationship
    dropout1 = np.clip(
        dropout1,
        0.10,
        0.50
    )

    dropout2 = np.clip(
        dropout2,
        0.10,
        0.40
    )

    dropout3 = np.clip(
        dropout3,
        0.05,
        0.30
    )

    model = build_aja_endnn(
        learning_rate,
        dropout1,
        dropout2,
        dropout3
    )

    callback = EarlyStopping(
        monitor="val_loss",
        patience=4,
        restore_best_weights=True,
        verbose=0
    )

    model.fit(
        X_aja_train,
        y_aja_train,
        validation_data=(
            X_aja_val,
            y_aja_val
        ),
        epochs=20,
        batch_size=32,
        verbose=0,
        callbacks=[callback]
    )

    _, accuracy = model.evaluate(
        X_aja_val,
        y_aja_val,
        verbose=0
    )

    tf.keras.backend.clear_session()

    return 1.0 - accuracy


# ------------------------------------------------------------
# Adaptive Jaya
# ------------------------------------------------------------

def adaptive_jaya(
    objective,
    lower_bounds,
    upper_bounds,
    population_size=5,
    iterations=3
):

    dimension = len(
        lower_bounds
    )

    population = np.random.uniform(
        lower_bounds,
        upper_bounds,
        size=(
            population_size,
            dimension
        )
    )

    fitness = np.array([
        objective(individual)
        for individual in population
    ])

    best_index = np.argmin(
        fitness
    )

    best_solution = population[
        best_index
    ].copy()

    best_fitness = fitness[
        best_index
    ]

    for iteration in range(
        iterations
    ):

        worst_index = np.argmax(
            fitness
        )

        worst_solution = population[
            worst_index
        ]

        for i in range(
            population_size
        ):

            r1 = np.random.rand(
                dimension
            )

            r2 = np.random.rand(
                dimension
            )

            candidate = (
                population[i]
                +
                r1 * (
                    best_solution
                    -
                    np.abs(
                        population[i]
                    )
                )
                -
                r2 * (
                    worst_solution
                    -
                    np.abs(
                        population[i]
                    )
                )
            )

            candidate = np.clip(
                candidate,
                lower_bounds,
                upper_bounds
            )

            candidate_fitness = objective(
                candidate
            )

            if candidate_fitness < fitness[i]:

                population[i] = candidate

                fitness[i] = candidate_fitness

                if candidate_fitness < best_fitness:

                    best_solution = candidate.copy()

                    best_fitness = candidate_fitness

        print(
            f"AJA Iteration "
            f"{iteration + 1}/{iterations} | "
            f"Best Validation Accuracy = "
            f"{(1 - best_fitness) * 100:.4f}%"
        )

    return best_solution, best_fitness


lower_bounds = np.array([
    0.0001,   # learning rate
    0.10,     # dropout 1
    0.10,     # dropout 2
    0.05      # dropout 3
])

upper_bounds = np.array([
    0.0050,
    0.50,
    0.40,
    0.30
])

best_aja_solution, best_aja_fitness = adaptive_jaya(
    aja_objective,
    lower_bounds,
    upper_bounds,
    population_size=5,
    iterations=3
)

best_lr = best_aja_solution[0]
best_dropout1 = best_aja_solution[1]
best_dropout2 = best_aja_solution[2]
best_dropout3 = best_aja_solution[3]

print("\n" + "-" * 80)
print("BEST AJA PARAMETERS")
print("-" * 80)

print(
    f"Learning Rate : {best_lr:.8f}"
)

print(
    f"Dropout 1     : {best_dropout1:.4f}"
)

print(
    f"Dropout 2     : {best_dropout2:.4f}"
)

print(
    f"Dropout 3     : {best_dropout3:.4f}"
)

# ============================================================
# 16. TRAIN FINAL AJA-OPTIMIZED EnDNN
# ============================================================

print("\n" + "=" * 100)
print("STEP 12: AJA-OPTIMIZED EnDNN")
print("=" * 100)

aja_endnn_model = build_aja_endnn(
    best_lr,
    best_dropout1,
    best_dropout2,
    best_dropout3
)

aja_endnn_model.fit(
    X_train_pca,
    y_train,
    validation_split=0.20,
    epochs=50,
    batch_size=32,
    shuffle=True,
    callbacks=[
        EarlyStopping(
            monitor="val_loss",
            patience=10,
            restore_best_weights=True,
            verbose=0
        ),
        ReduceLROnPlateau(
            monitor="val_loss",
            factor=0.5,
            patience=5,
            min_lr=1e-6,
            verbose=0
        )
    ],
    verbose=0
)

aja_prob = aja_endnn_model.predict(
    X_test_pca,
    batch_size=32,
    verbose=0
)

aja_pred = np.argmax(
    aja_prob,
    axis=1
)

aja_metrics = calculate_metrics(
    y_test,
    aja_pred
)

print("\nAJA-EnDNN Results:")

for key, value in aja_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 17. EDGE COMPUTING INTEGRATION
# ============================================================

print("\n" + "=" * 100)
print("STEP 13: EDGE COMPUTING INTEGRATION")
print("=" * 100)

# ------------------------------------------------------------
# Edge inference uses the trained AJA-EnDNN model.
#
# The edge layer performs:
#   - local input normalization
#   - local PCA transformation
#   - local model inference
#
# No cloud-side prediction is required.
# ------------------------------------------------------------

def edge_inference(
    X_input
):

    # Local preprocessing
    X_scaled = scaler.transform(
        X_input
    )

    # Local PCA
    X_edge_pca = pca.transform(
        X_scaled
    )

    # Local inference
    probabilities = aja_endnn_model.predict(
        X_edge_pca.astype("float32"),
        batch_size=32,
        verbose=0
    )

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    return predictions


edge_pred = edge_inference(
    X_test_raw
)

edge_metrics = calculate_metrics(
    y_test,
    edge_pred
)

print("\nEdge Computing Results:")

for key, value in edge_metrics.items():
    print(
        f"{key}: {value:.4f}"
    )

# ============================================================
# 18. PROPOSED AJA-EnDNN
# ============================================================

print("\n" + "=" * 100)
print("STEP 14: AJA-EnDNN [PROPOSED]")
print("=" * 100)

# ------------------------------------------------------------
# The proposed model combines:
#
# Data preprocessing
#        +
# PCA feature extraction
#        +
# EnDNN
#        +
# AJA optimization
#        +
# Edge inference
#
# Therefore the edge prediction above is the final proposed
# prediction.
# ------------------------------------------------------------

proposed_metrics = edge_metrics.copy()

# ============================================================
# 19. MODEL VARIANT COMPARISON
# ============================================================

print("\n" + "=" * 100)
print("MODEL VARIANT COMPARISON")
print("=" * 100)

results = [

    {
        "Model Variant":
            "Baseline DNN",

        "Description":
            "Conventional DNN without preprocessing or optimization",

        **baseline_metrics
    },

    {
        "Model Variant":
            "+ Data Preprocessing",

        "Description":
            "DNN with data cleaning and normalization",

        **preprocessing_metrics
    },

    {
        "Model Variant":
            "+ PCA Feature Extraction",

        "Description":
            "DNN with dimensionality reduction using PCA",

        **pca_metrics
    },

    {
        "Model Variant":
            "+ EnDNN",

        "Description":
            "Enhanced DNN with enriched feature learning and nonlinear representation",

        **endnn_metrics
    },

    {
        "Model Variant":
            "+ AJA Optimization",

        "Description":
            "EnDNN integrated with Adaptive Jaya hyperparameter optimization",

        **aja_metrics
    },

    {
        "Model Variant":
            "+ Edge Computing Integration",

        "Description":
            "Real-time monitoring with edge-based processing optimization",

        **edge_metrics
    },

    {
        "Model Variant":
            "AJA-EnDNN [Proposed]",

        "Description":
            "Fully integrated AI- and edge-computing-based labor education monitoring framework",

        **proposed_metrics
    }
]

comparison_df = pd.DataFrame(
    results
)

# ============================================================
# 20. FORMAT RESULTS
# ============================================================

metric_columns = [
    "Accuracy (%)",
    "Precision (%)",
    "Recall (%)",
    "F1-Score (%)",
    "Sensitivity (%)",
    "Specificity (%)",
    "MCC"
]

for col in metric_columns:

    comparison_df[col] = comparison_df[col].astype(
        float
    )

# ============================================================
# 21. DISPLAY DATAFRAME
# ============================================================

print("\n")

print(
    comparison_df.to_string(
        index=False,
        formatters={
            "Accuracy (%)":
                lambda x: f"{x:.2f}",

            "Precision (%)":
                lambda x: f"{x:.2f}",

            "Recall (%)":
                lambda x: f"{x:.2f}",

            "F1-Score (%)":
                lambda x: f"{x:.2f}",

            "Sensitivity (%)":
                lambda x: f"{x:.2f}",

            "Specificity (%)":
                lambda x: f"{x:.2f}",

            "MCC":
                lambda x: f"{x:.4f}"
        }
    )
)

# ============================================================
# 22. FINAL JOURNAL-STYLE TABLE
# ============================================================

print("\n")
print("=" * 140)

header = (
    f"{'Model Variant':<32}"
    f"{'Accuracy (%)':>14}"
    f"{'Precision (%)':>15}"
    f"{'Recall (%)':>12}"
    f"{'F1-Score (%)':>14}"
    f"{'Sensitivity (%)':>17}"
    f"{'Specificity (%)':>17}"
    f"{'MCC':>10}"
)

print(header)

print("-" * 140)

for _, row in comparison_df.iterrows():

    print(
        f"{row['Model Variant']:<32}"
        f"{row['Accuracy (%)']:>14.2f}"
        f"{row['Precision (%)']:>15.2f}"
        f"{row['Recall (%)']:>12.2f}"
        f"{row['F1-Score (%)']:>14.2f}"
        f"{row['Sensitivity (%)']:>17.2f}"
        f"{row['Specificity (%)']:>17.2f}"
        f"{row['MCC']:>10.4f}"
    )

print("=" * 140)

# ============================================================
# 23. CHECK FOR NAN VALUES
# ============================================================

print("\n" + "=" * 100)
print("RESULT VALIDATION")
print("=" * 100)

print("\nNaN Values in Comparison Table:")

print(
    comparison_df[
        metric_columns
    ].isna().sum()
)

print("\nInfinite Values:")

print(
    np.isinf(
        comparison_df[
            metric_columns
        ].values.astype(float)
    ).sum()
)

print("\n" + "=" * 100)
print("FULL ABLATION STUDY COMPLETED")
print("=" * 100)
```
